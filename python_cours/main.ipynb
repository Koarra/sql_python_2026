```python
#### FOR LOOP: itere a travers une liste pour appliquer un critere (operators, classification, peu importe)
liste_de_pays_a_voyager_2026 = ["Bresil", "Espagne", "Inde", "Senegal"]

continent_amerique = ["Bresil", "USA", "Mexique"]
continent_europe = ["Espagne", "portugal", "France", "italie"]
continent_afrique = ["mali", "Cote d'ivoir", "Senegal", "egypt","Congo"]
continent_asie = ["Inde", "chine", "coree du sud"]

# Dans quelles continent je vais voyager en 2026?
continent_visite = []

for pays in liste_de_pays_a_voyager_2026:
    if pays in continent_amerique:
        continent_visite.append("amerique")
    elif pays in continent_europe:
        continent_visite.append("europe")
    elif pays in continent_afrique:
        continent_visite.append("afrique")
    elif pays in continent_asie:
        continent_visite.append("asie")
```
