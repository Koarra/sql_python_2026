```python
### une fonction permet d'automatiser une tache. On donne un nom a une fonction, et on peut faire appel
### a la fonction lorsqu'on veut, en lui fournissant a chaque fois des inputs.
```

```python
# Calculer la moyenne general d'un seul etudiant 

# sans une fonction
note_de_math = 15
note_de_francais = 12
note_de_machine_learning = 14

somme_des_notes = note_de_math + note_de_francais + note_de_machine_learning
nombre_de_matieres = 3

moyenne_generale = somme_des_notes / nombre_de_matieres

print("La moyenne generale de l'etudiant est de:", moyenne_generale)
```

```python
# avec une fonction 

def calculer_moyenne_eleve(note1, note2, note3, nb_de_matieres):
    somme_des_notes = note1 + note2 + note3
    moyenne_generale = somme_des_notes / nb_de_matieres

    return moyenne_generale
```

```python
note_de_math_fanny = 18
note_de_francais_fanny = 17
note_de_machine_learning_fanny = 19

# Je vais fair appel a ma fonction creer pour cette tache
# obligation de mettre les memes nombre d'elements en inputs, ici 4 elements ont ete definis dans la fonction
#  
moyenne_de_fanny = calculer_moyenne_eleve(note_de_math_fanny, note_de_francais_fanny, note_de_machine_learning_fanny, 3)
print(moyenne_de_fanny)
```

```python

def nettoyer_text(input_text):
    caracteres_a_supprimer = ["#", "$", "%", "!',", "@", "?!)*()", "*", "&"]
    texte_nettoye = input_text

    for caractere in caracteres_a_supprimer:
        texte_nettoye = texte_nettoye.replace(caractere, "")

    return texte_nettoye

```

```python
# creer une troisieme dataframe a partir des 2 existantes, 
# pour avoir une colonne total_weight et une colonne weight 
# # il faut que ca ait la meme structure que sells_df + les deux nouvelles colonnes



# --> je copie la table sells a laquelle je vais juste ajouter 2 nouvelles colonnes 
# sells_df + 2 nouvelles colonnes 
#     - une colonne qui existe deja dans table product
#     - une nouvelle colonne a devoir creer via calcul quantite * weight
```

```python
# Left JOIN choisi pour repondre a cette requete 
# exemple de jointure 
# nouvelle_df = premiere_table.merge(deuxieme_table, on="colonne_commune", how="left")

# notre probleme est le suivant: les colonnes communes entre les deux tables sont product et product_information 
# !!! elles ne portent pas le meme nom dans les deux tables
# 2 solutions:
    # solution 1: renommer une des deux colonnes pour qu'elles aient le meme nom
    # solution 2: utiliser les parametres left_on et right_on dans la fonction merge
sells_enriched_df = sells_df.merge(
    product_df[["product", "weight"]],
    left_on = " product_information",
    right_on = "product",
    how = "left"
)
```

```python
sells_enriched_df
```

```python
# creer ma nouvelle colonne: total_weight qui va etre egale a quantity * weight
sells_enriched_df["total_weight"] = sells_enriched_df[" quantity"] * sells_enriched_df["weight"]

sells_enriched_df
```
